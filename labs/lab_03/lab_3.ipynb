```python
import pandas as pd
import numpy as np
from collections import defaultdict

# Load and clean data
df = pd.read_csv('../data/lab01_train.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp').drop_duplicates('id', keep='last')
numeric_cols = ['course_2', 'course_3', 'is_mi_student', 'is_ml_student', 'percentile', 'rating']
df[numeric_cols] = df[numeric_cols].fillna(0)
string_cols = ['fall_1', 'fall_2', 'fall_3', 'spring_1', 'spring_2', 'spring_3', 'blended', 'is_first_time']
df[string_cols] = df[string_cols].fillna('')

# Create fall_course_num
df['fall_course_num'] = 1
mop_ti = [61, 62, 63]
df.loc[df['course_3'].isin(mop_ti), 'fall_course_num'] = 2
df.loc[(df['course_2'].notna()) & (df['is_mi_student'] == 1), 'fall_course_num'] = 2

# Fall courses
fall_cols = ['fall_1', 'fall_2', 'fall_3']
all_fall = df[fall_cols].stack().dropna().unique()
fall_courses = sorted([c for c in all_fall if c != ''])

# Capacities
capacities = {c: 30 for c in fall_courses}
capacities['Statistical Learning Theory'] = 60
capacities['Высокопроизводительные вычисления'] = 60
capacities['Анализ неструктурированных данных'] = np.inf

# Dicts
percentiles = df.set_index('id')['percentile'].to_dict()
num_fall = df.set_index('id')['fall_course_num'].to_dict()

# Assigned
assigned = defaultdict(list)

# Remaining capacity
rem_cap = capacities.copy()

# Wave 1: First priority
wave1_cands = defaultdict(list)
for _, row in df.iterrows():
    id_ = row['id']
    num = num_fall[id_]
    first_prio = []
    if row['fall_1'] != '':
        first_prio.append(row['fall_1'])
    if num == 2 and row['fall_2'] != '':
        first_prio.append(row['fall_2'])
    for course in set(first_prio):
        wave1_cands[course].append((percentiles[id_], id_))

enrolled_wave1 = defaultdict(list)
for course in fall_courses:
    if course in wave1_cands:
        cands = sorted(wave1_cands[course], key=lambda x: (x[0], x[1]))  # low pct first, then low id
        take = min(len(cands), rem_cap[course])
        taken = [cand[1] for cand in cands[:take]]
        for id_ in taken:
            if len(assigned[id_]) < num_fall[id_]:
                assigned[id_].append(course)
                enrolled_wave1[course].append(id_)
        rem_cap[course] -= take

# is_first_place and is_last_place
is_first = {}
is_last = {}
for course, studs in enrolled_wave1.items():
    if studs:
        pcts = [percentiles[s] for s in studs]
        best_p = min(pcts)  # low better
        worst_p = max(pcts)
        for s in studs:
            if percentiles[s] == best_p:
                is_first[s] = True
            if percentiles[s] == worst_p:
                is_last[s] = True

df['is_first_place'] = df['id'].map(is_first)
df['is_last_place'] = df['id'].map(is_last)

# Wave 2: Second priority
wave2_cands = defaultdict(list)
for _, row in df.iterrows():
    id_ = row['id']
    if len(assigned[id_]) < num_fall[id_]:
        num = num_fall[id_]
        second_prio = []
        if num == 1 and row['fall_2'] != '':
            second_prio.append(row['fall_2'])
        if num == 2 and row['fall_3'] != '':
            second_prio.append(row['fall_3'])
        for course in set(second_prio):
            wave2_cands[course].append((percentiles[id_], id_))

enrolled_wave2 = defaultdict(list)
for course in fall_courses:
    if course in wave2_cands:
        cands = sorted(wave2_cands[course], key=lambda x: (x[0], x[1]))
        take = min(len(cands), rem_cap[course])
        taken = [cand[1] for cand in cands[:take]]
        for id_ in taken:
            if len(assigned[id_]) < num_fall[id_]:
                assigned[id_].append(course)
                enrolled_wave2[course].append(id_)
        rem_cap[course] -= take

# Wave 3: Third priority (only for num=1)
wave3_cands = defaultdict(list)
for _, row in df.iterrows():
    id_ = row['id']
    if num_fall[id_] == 1 and len(assigned[id_]) < 1 and row['fall_3'] != '':
        wave3_cands[row['fall_3']].append((percentiles[id_], id_))

enrolled_wave3 = defaultdict(list)
for course in fall_courses:
    if course in wave3_cands:
        cands = sorted(wave3_cands[course], key=lambda x: (x[0], x[1]))
        take = min(len(cands), rem_cap[course])
        taken = [cand[1] for cand in cands[:take]]
        for id_ in taken:
            if len(assigned[id_]) < num_fall[id_]:
                assigned[id_].append(course)
                enrolled_wave3[course].append(id_)
        rem_cap[course] -= take

# Create res_fall.csv
res_df = pd.DataFrame({'ID': df['id']})
for _, row in df.iterrows():
    id_ = row['id']
    ass_courses = assigned[id_]
    num = num_fall[id_]
    if len(ass_courses) < num:
        ass_courses += ['???'] * (num - len(ass_courses))
    if num == 1:
        res_df.loc[res_df['ID'] == id_, 'course1'] = ass_courses[0] if len(ass_courses) > 0 else '???'
        res_df.loc[res_df['ID'] == id_, 'course2'] = '-'
    else:
        res_df.loc[res_df['ID'] == id_, 'course1'] = ass_courses[0] if len(ass_courses) > 0 else '???'
        res_df.loc[res_df['ID'] == id_, 'course2'] = ass_courses[1] if len(ass_courses) > 1 else '???'

res_df.to_csv('res_fall.csv', index=False)

# For the assert (run this after loading check_df)
# check_df = pd.read_csv('./data/lab01_train_check.csv')
# your_first_ids = df[df['is_first_place'].notna()]['id'].sort_values().values
# check_first_ids = check_df[check_df['is_first_place'].notna()]['id'].sort_values().values
# assert np.array_equal(your_first_ids, check_first_ids)
# Similar for last_place

print("Code executed. File 'res_fall.csv' saved. Run the assert separately with check_df loaded.")
```
